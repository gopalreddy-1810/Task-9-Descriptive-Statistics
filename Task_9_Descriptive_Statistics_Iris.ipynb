```python
df = pd.read_csv("Iris.csv")

df.head()
```

```python
numeric_data = df[
    ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
]

numeric_data.head()
```

```python
print("Mean:")
print(numeric_data.mean())

print("\nMedian:")
print(numeric_data.median())

print("\nMode:")
print(numeric_data.mode().iloc[0])
```

```python
print("Standard Deviation:")
print(numeric_data.std())

print("\n25th Percentile:")
print(numeric_data.quantile(0.25))

print("\n50th Percentile:")
print(numeric_data.quantile(0.50))

print("\n75th Percentile:")
print(numeric_data.quantile(0.75))
```

```python
summary = pd.DataFrame({
    "Mean": numeric_data.mean(),
    "Median": numeric_data.median(),
    "Mode": numeric_data.mode().iloc[0],
    "Standard Deviation": numeric_data.std(),
    "25th Percentile": numeric_data.quantile(0.25),
    "50th Percentile": numeric_data.quantile(0.50),
    "75th Percentile": numeric_data.quantile(0.75)
})

summary
```

```python
print("INTERPRETATION")
print("1. Mean and median help describe the central tendency of the Iris measurements.")
print("2. Mode shows the most frequently occurring value in each numeric column.")
print("3. Standard deviation shows how much the values vary from the mean.")
print("4. The 25th, 50th, and 75th percentiles describe the distribution of the data.")
print("5. Overall, descriptive statistics summarize the center and spread of the four Iris measurements.")
```
